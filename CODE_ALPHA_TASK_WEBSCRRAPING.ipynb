```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

print("Libraries imported successfully!")
```

```python
url = "https://books.toscrape.com/"
response = requests.get(url,timeout=30)
print(response.status_code)
```

```python
soup = BeautifulSoup(response.text,
"html.parser")

print("Website HTML loaded successfully!")
print(soup.title.text)
```

```python

books = []

for product in soup.select("article.product_pod"):
    books.append({
        "Title": product.h3.a["title"],
        "Price": product.select_one(".price_color").get_text(strip=True),
        "Availability": product.select_one(".availability").get_text(strip=True)
    })

print("Books collected:", len(books))
print(books[:5])
```

```python
df = pd.DataFrame(books)

print(df.head())
```

```python
df.to_csv("books_data.csv",
index=False)

print("Dataset saved successfully!")
```
