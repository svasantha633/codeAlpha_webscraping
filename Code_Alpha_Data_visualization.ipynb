```python
import pandas as pd

df = pd.read_csv("books_data.csv")

print("Dataset shape:", df.shape)
print(df.head())
print(df.columns.tolist())
```

```python
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("Â£", "", regex=False)
    .str.replace("£", "", regex=False)
    .astype(float)
)

print(df[["Title", "Price"]].head())
```

```python
import matplotlib.pyplot as plt

top_10 = df.nlargest(10, "Price")

plt.figure(figsize=(12, 6))
plt.barh(top_10["Title"], top_10["Price"], color="blue")
plt.title("Top 10 Most Expensive Books")
plt.xlabel("Price (£)")
plt.ylabel("Book Title")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["Price"].dropna(), bins=10, color="skyblue", edgecolor="black")

plt.title("Distribution of Book Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

availability_counts = df["Availability"].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(
    availability_counts.index.astype(str),
    availability_counts.values,
    color="skyblue",
    edgecolor="black"
)

plt.title("Book Availability Count")
plt.xlabel("Availability")
plt.ylabel("Number of Books")
plt.tight_layout()
plt.show()
```
